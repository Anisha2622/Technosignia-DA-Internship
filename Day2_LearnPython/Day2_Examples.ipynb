```python
# Program 1: Variables and data types

student_name = "Anisha"       # string (text)
age = 23                      # integer (whole number)
percentage = 86.5             # float (decimal number)
is_learning_python = True     # Boolean (True or False)

print("Student name:", student_name)
print("Age:", age)
print("Percentage:", percentage)
print("Learning Python:", is_learning_python)

print("\\nData types:")
print(type(student_name))
print(type(age))
print(type(percentage))
print(type(is_learning_python))
```

```python
# Program 2: Arithmetic operators

first_number = 20
second_number = 6

print("Addition:", first_number + second_number)
print("Subtraction:", first_number - second_number)
print("Multiplication:", first_number * second_number)
print("Division:", first_number / second_number)
print("Whole-number division:", first_number // second_number)
print("Remainder:", first_number % second_number)
```

```python
# Program 3: Comparison and logical operators

marks = 78
attendance = 85

print("Marks are at least 40:", marks >= 40)
print("Marks are exactly 100:", marks == 100)
print("Marks are below 35:", marks < 35)

passed_exam = marks >= 40
good_attendance = attendance >= 75

print("Passed and has good attendance:", passed_exam and good_attendance)
print("Passed or has perfect attendance:", passed_exam or attendance == 100)
print("Did not pass:", not passed_exam)
```

```python
# Program 4: Input and type conversion

name = input("Enter your name: ")
age = int(input("Enter your age: "))

print("Hello,", name)
print("Next year, you will be", age + 1)
```

```python
# Program 5: Calculate percentage

total_marks = 500
marks_obtained = 425

percentage = (marks_obtained / total_marks) * 100

print("Marks obtained:", marks_obtained)
print("Total marks:", total_marks)
print("Percentage:", percentage, "%")
```

```python
# Program 6: Calculate sales total

product_name = "Notebook"
quantity_sold = 12
price_per_item = 40.0

sales_total = quantity_sold * price_per_item

print("Product:", product_name)
print("Quantity sold:", quantity_sold)
print("Price per item: Rs.", price_per_item)
print("Total sales: Rs.", sales_total)
```

```python
# Program 7: Calculate discount

original_price = 1500.0
discount_percent = 10.0

discount_amount = original_price * discount_percent / 100
final_price = original_price - discount_amount

print("Original price: Rs.", original_price)
print("Discount:", discount_percent, "%")
print("Discount amount: Rs.", discount_amount)
print("Final price: Rs.", final_price)
```

```python
# Program 8: Calculate average marks

mark1 = 75
mark2 = 82
mark3 = 91
mark4 = 68
mark5 = 84

average = (mark1 + mark2 + mark3 + mark4 + mark5) / 5

print("Average marks:", average)
```
