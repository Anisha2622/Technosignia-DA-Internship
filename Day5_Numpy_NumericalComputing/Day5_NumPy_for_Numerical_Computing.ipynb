# Day 5 — NumPy for Numerical Computing

**Task:** Learn numerical calculations using NumPy.

NumPy is a Python library that helps us work with numbers and arrays efficiently. It is commonly used in data analysis, statistics, and machine learning. Run each code cell from top to bottom in Google Colab or Jupyter Notebook.


## 1. Import NumPy

NumPy is often imported with the short name `np`. Google Colab usually has NumPy already installed. If you are running Python locally and it is missing, install it in your terminal with `python -m pip install numpy`.


```python
# Import the NumPy library
import numpy as np

print("NumPy version:", np.__version__)
```

## 2. Create one-dimensional and multidimensional arrays

An array stores values in an organised structure. A one-dimensional array is like a single row of values. A two-dimensional array looks like a table with rows and columns.


```python
# Create a one-dimensional array
daily_sales = np.array([120, 250, 180, 90, 300])
print("Daily sales:", daily_sales)
print("Number of dimensions:", daily_sales.ndim)
print("Shape:", daily_sales.shape)

# Create a two-dimensional array
weekly_sales = np.array([
    [120, 250, 180],
    [90, 300, 210]
])
print("\nWeekly sales data:")
print(weekly_sales)
print("Number of dimensions:", weekly_sales.ndim)
print("Shape (rows, columns):", weekly_sales.shape)
```

## 3. Indexing and slicing

Indexing gets one value from an array. Python uses zero-based indexing, so the first value is at position `0`. Slicing selects a range of values; the ending position is not included.


```python
# Indexing and slicing
numbers = np.array([10, 20, 30, 40, 50, 60])

print("First value:", numbers[0])
print("Third value:", numbers[2])
print("Values from index 1 to 3:", numbers[1:4])
print("Last two values:", numbers[-2:])
```

## 4. Filtering values

Filtering helps select only values that match a condition. The condition creates True/False results, and using it inside the array returns the matching values.


```python
# Select sales values greater than 150
sales = np.array([120, 250, 180, 90, 300, 140])
high_sales = sales[sales > 150]

print("All sales:", sales)
print("Sales greater than 150:", high_sales)
```

## 5. Reshape an array

Reshaping changes the arrangement of values without changing the number of values. For example, six values can be arranged into two rows and three columns.


```python
# Reshape six values into 2 rows and 3 columns
values = np.array([1, 2, 3, 4, 5, 6])
matrix = values.reshape(2, 3)

print("Original array:", values)
print("Reshaped array:")
print(matrix)
```

## 6. Calculate sum, mean, minimum, maximum, and standard deviation

NumPy provides ready-made functions for common numerical summaries. The standard deviation describes how spread out values are around the mean. NumPy's default standard deviation uses the population formula (`ddof=0`).


```python
# Basic numerical summaries
marks = np.array([70, 85, 90, 65, 80])

print("Marks:", marks)
print("Sum:", np.sum(marks))
print("Mean (average):", np.mean(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Standard deviation:", np.std(marks))
```

## 7. Compare basic Python calculations with NumPy

With a regular Python list, calculations such as adding a fixed amount to every value usually need a loop or a list comprehension. NumPy can apply the operation to the whole array directly. This is called vectorized operation.


```python
# Python list calculation
python_numbers = [10, 20, 30, 40]
python_plus_five = [number + 5 for number in python_numbers]
print("Python list + 5:", python_plus_five)

# NumPy array calculation
numpy_numbers = np.array([10, 20, 30, 40])
numpy_plus_five = numpy_numbers + 5
print("NumPy array + 5:", numpy_plus_five)
```

## 8. Small practical example: sales analysis

This example uses NumPy to calculate the average daily sales and identify days with sales above the average.


```python
# Practical example: daily sales
daily_sales = np.array([1200, 1500, 900, 1800, 1600, 1100, 2000])

average_sales = np.mean(daily_sales)
above_average = daily_sales[daily_sales > average_sales]

print("Daily sales:", daily_sales)
print("Average daily sales:", average_sales)
print("Days with sales above average:", above_average)
print("Highest daily sales:", np.max(daily_sales))
```
