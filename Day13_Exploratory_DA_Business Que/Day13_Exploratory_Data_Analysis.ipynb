# Day 13 — Exploratory Data Analysis and Business Questions

**Task:** Learn to convert business problems into analytical questions.

This practical uses fictional sales data from January to May 2026. The goal is not just to make charts, but to ask useful business questions, select the right metrics, analyse the data, and explain what the results might mean. Run the notebook cells from top to bottom.

**Dataset:** `day13_eda_sales_data.csv`


## 1. Load and inspect the dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("day13_eda_sales_data.csv", parse_dates=["Order Date"])
display(df.head())
print("Dataset shape (rows, columns):", df.shape)
print("\nColumn data types:")
print(df.dtypes)
print("\nMissing values per column:")
print(df.isna().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

## 2. Prepare useful metrics

```python
df["Sales (₹)"] = df["Quantity"] * df["Unit Price (₹)"]
df["Cost (₹)"] = df["Quantity"] * df["Unit Cost (₹)"]
df["Profit (₹)"] = df["Sales (₹)"] - df["Cost (₹)"]
df["Month"] = df["Order Date"].dt.to_period("M").astype(str)

total_sales = df["Sales (₹)"].sum()
total_profit = df["Profit (₹)"].sum()
total_orders = df["Order ID"].nunique()
units_sold = df["Quantity"].sum()
average_order_value = df["Sales (₹)"].mean()

print(f"Total sales: ₹{total_sales:,.0f}")
print(f"Total profit: ₹{total_profit:,.0f}")
print(f"Number of orders: {total_orders}")
print(f"Units sold: {units_sold}")
print(f"Average sales per order record: ₹{average_order_value:,.2f}")
```

## 3. Define five business questions and the metrics

1. **Which category generates the most sales?** Metric: total sales by category.
2. **Which city contributes the most sales?** Metric: total sales by city.
3. **Are sales increasing over time?** Metric: monthly total sales.
4. **Which products generate the most sales?** Metric: total sales by product.
5. **Which sales channel contributes more revenue?** Metric: total sales by channel.

The code below calculates the metrics needed to answer these questions.


```python
sales_by_category = df.groupby("Category")["Sales (₹)"].sum().sort_values(ascending=False)
sales_by_city = df.groupby("City")["Sales (₹)"].sum().sort_values(ascending=False)
sales_by_month = df.groupby("Month")["Sales (₹)"].sum().sort_index()
sales_by_product = df.groupby("Product")["Sales (₹)"].sum().sort_values(ascending=False)
sales_by_channel = df.groupby("Sales Channel")["Sales (₹)"].sum().sort_values(ascending=False)

print("Sales by category:\n", sales_by_category)
print("\nSales by city:\n", sales_by_city)
print("\nSales by month:\n", sales_by_month)
print("\nSales by product:\n", sales_by_product)
print("\nSales by channel:\n", sales_by_channel)
```

## 4. Visualize the findings

```python
plt.figure(figsize=(8, 4))
sales_by_category.plot(kind="bar", title="Total Sales by Category", xlabel="Category", ylabel="Sales (₹)")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 4))
sales_by_month.plot(kind="line", marker="o", title="Monthly Sales Trend", xlabel="Month", ylabel="Sales (₹)")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 4))
sales_by_city.plot(kind="bar", title="Total Sales by City", xlabel="City", ylabel="Sales (₹)")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 4))
sales_by_channel.plot(kind="bar", title="Sales by Channel", xlabel="Sales Channel", ylabel="Sales (₹)")
plt.tight_layout()
plt.show()
```

## 5. Map business questions to metrics and findings

```python
findings = pd.DataFrame([
    {"Business question": "Which category generates the most sales?",
     "Metric": "Total sales by category",
     "Finding": f"{sales_by_category.index[0]} leads with ₹{sales_by_category.iloc[0]:,.0f} in sales."},
    {"Business question": "Which city contributes the most sales?",
     "Metric": "Total sales by city",
     "Finding": f"{sales_by_city.index[0]} leads with ₹{sales_by_city.iloc[0]:,.0f} in sales."},
    {"Business question": "Are sales increasing over time?",
     "Metric": "Monthly total sales",
     "Finding": f"Sales increased from ₹{sales_by_month.iloc[0]:,.0f} in {sales_by_month.index[0]} to ₹{sales_by_month.iloc[-1]:,.0f} in {sales_by_month.index[-1]}."},
    {"Business question": "Which product generates the most sales?",
     "Metric": "Total sales by product",
     "Finding": f"{sales_by_product.index[0]} leads with ₹{sales_by_product.iloc[0]:,.0f} in sales."},
    {"Business question": "Which channel contributes more revenue?",
     "Metric": "Total sales by channel",
     "Finding": f"{sales_by_channel.index[0]} contributes ₹{sales_by_channel.iloc[0]:,.0f}, compared with ₹{sales_by_channel.iloc[1]:,.0f} for {sales_by_channel.index[1]}."}
])
display(findings)
```

## 6. Findings and limitations

### Main findings
- Total sales and profit summarise the overall business performance in this sample.
- Category and city comparisons help identify where sales are concentrated.
- Monthly sales shows whether revenue is moving up or down during the available period.
- Product and channel comparisons can help a business decide where to investigate further.

### Limitations
- This is a small, fictional dataset created for practice, not real company data.
- It covers only January to May 2026, so it cannot establish a long-term seasonal trend.
- Sales totals do not explain why customers bought products.
- A relationship between two measures does not automatically mean one caused the other.
- Before making a real business decision, an analyst should confirm data quality and ask stakeholders whether other factors need to be considered.
